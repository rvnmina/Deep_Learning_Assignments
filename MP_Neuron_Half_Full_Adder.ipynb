## Question 1 (10 marks)

Design a half adder circuit using the minimum possible number of MP neurons.

```python
import numpy as np

def mp_neuron(inputs, threshold, inhibitory_indices=None): #MP neuron with inhibitory inputs
    if inhibitory_indices is None:
        inhibitory_indices = []

    for idx in inhibitory_indices:
        if inputs[idx] == 1:
            return 0

    return 1 if np.sum(inputs) >= threshold else 0
```

```python
def XOR(a, b):
    n1 = mp_neuron(np.array([a, b]), threshold=1, inhibitory_indices=[1]) #a AND NOT b

    n2 = mp_neuron(np.array([a, b]), threshold=1, inhibitory_indices=[0]) #NOT a AND b

    return mp_neuron(np.array([n1, n2]), threshold=1)  #OR above two
```

```python
def half_adder(a, b):
    sum_bit = XOR(a, b)
    carry = mp_neuron(np.array([a, b]), threshold=2)
    return sum_bit, carry
```

## Question 2 (2 marks)

Test the half adder that you designed on all possible input combinations.

```python
inputs = [(0,0), (0,1), (1,0), (1,1)]

sum_outputs = []
carry_outputs = []

for a, b in inputs:
    s, c = half_adder(a, b)
    sum_outputs.append(s)
    carry_outputs.append(c)

print("Inputs :", inputs)
print("\nGround Truth:")
print("Sum   :", [0, 1, 1, 0])
print("Carry :", [0, 0, 0, 1])

print("\nHalf Adder Outputs:")
print("Sum   :", sum_outputs)
print("Carry :", carry_outputs)
```

## Question 3 (6 marks)

Using two half adders and an OR gate, design a full adder.

```python
def full_adder(a, b, cin):
    s1, c1 = half_adder(a, b) #First half adder

    sum_bit, c2 = half_adder(s1, cin) #Second half adder

    cout = mp_neuron([c1, c2], 1) #OR gate for carry

    return sum_bit, cout
```
